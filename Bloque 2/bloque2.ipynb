```python
# Imports
import json
import numpy as np
from codigo_base import *

# Variables globales
rng = np.random.default_rng(4)
```

```python
R = 2000
N1 = 100
N2 = 20
n = 100

def medir_comparaciones_quick_sort(N):
    medianas = []
    medias = []
    for i in range(0, R):
        comparacionesQS = []
        for r in range(N):
            entrada = permutacion_aleatoria(n, rng)
            _, comps = ALGORITMOS["quick_primero"](entrada)
            comparacionesQS.append(comps)
        mediana = np.median(comparacionesQS)
        media = np.mean(comparacionesQS)
        medianas.append(mediana)
        medias.append(media)

    media_teorica_Q100 = 2 * n * np.log(n) # se podria hacer mas preciso con la constante de Euler

    media_e1 = np.mean(medianas)
    varianza_muestral_e1 = np.var(medianas, ddof=1)
    sesgo_e1 = media_e1 - media_teorica_Q100
    ecm_e1 = varianza_muestral_e1 + sesgo_e1**2

    media_e2 = np.mean(medias)
    varianza_muestral_e2 = np.var(medias, ddof=1)
    sesgo_e2 = media_e2 - media_teorica_Q100
    ecm_e2 = varianza_muestral_e2 + sesgo_e2**2

    return sesgo_e1, varianza_muestral_e1, ecm_e1, sesgo_e2, varianza_muestral_e2, ecm_e2

print(medir_comparaciones_quick_sort(N1))
print(medir_comparaciones_quick_sort(N2))
```
